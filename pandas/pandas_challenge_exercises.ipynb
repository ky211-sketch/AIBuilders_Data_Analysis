## 3. Handling Missing Data

**Exercise 9.** Count the number of missing values in **each** column. (Hint: combine `.isnull()` with `.sum()`.)

```python
df.isnull().sum()
```

**Exercise 10.** Fill the missing `Age` values with the **median** age, and assign the result back to the `Age` column. Then confirm there are no missing ages left.

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age']
```

```python
df
```

```python
df['Age'].isnull().sum()
```

**Exercise 11.** The `Embarked` column has only 2 missing values. Fill them with the **most common** port (the mode). The `Cabin` column is mostly empty — **drop** that whole column instead.

*Tip: `df["Embarked"].mode()` returns a Series; take the first element with `[0]`. Use `df.drop(columns=...)` to remove a column.*

```python
mode_embarked = df["Embarked"].mode() #the most common value is S
mode_embarked
```

```python
df['Embarked'] = df['Embarked'].fillna(mode_embarked[0]) 
```

```python
df
```

```python
df['Embarked'].isnull().sum()
```

```python
df = df.drop('Cabin', axis=1)
df
```

```python
df.info()
```

**Exercise 12.** What was the overall **survival rate**? (Because `Survived` is 0/1, its mean equals the proportion who survived.)

```python
survival_rate = df['Survived'].mean() *100
survival_rate
```

**Exercise 13.** Group by `Sex` and find the survival rate for each. Did men or women survive at a higher rate?

```python
df1 = df.groupby(['Sex']).mean(numeric_only=True) 
df1
```

```python
survival_rate1 = df1['Survived'] *100
survival_rate1
# so female survive more than women
```

**Exercise 14.** Group by `Pclass` and find the survival rate for each class. Then group by **both** `Pclass` and `Sex` together.

## 5. Creating and Transforming Columns

**Exercise 16.** Create a new column `FamilySize` equal to `SibSp + Parch + 1` (the +1 counts the passenger themselves).

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FamilySize']
```

**Exercise 17.** Create a new column `IsAlone` that is `1` when `FamilySize == 1` and `0` otherwise.

*Tip: a boolean comparison gives True/False — convert it to integers with `.astype(int)`.*

```python
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
df['IsAlone']
#df['IsAlone'].astype(int) not saved as integer in the dataset
```

```python
#to check
df[['FamilySize', 'IsAlone']].head()
```

```python
df.head()
```

**Exercise 18.** Use `.apply()` with a custom function (or a `lambda`) to create an `AgeGroup` column that labels each passenger as `"Child"` if `Age < 18`, otherwise `"Adult"`.

```python
df['AgeGroup'] = df['Age'].apply(lambda x : 'child' if x<18 else 'Adult')
df['AgeGroup']
```

```python
df[['Age', 'AgeGroup']].head(10)
```

```python
df
```

**Exercise 19.** Using your new `AgeGroup` column, what was the survival rate for children vs adults? (Group by `AgeGroup` and take the mean of `Survived`.)

```python
survival_rate3 = df.groupby('AgeGroup')['Survived'].mean()
survival_rate3
```
